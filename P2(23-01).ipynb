```python
from matplotlib.image import imread
import  matplotlib.pyplot as plt
import cv2
import numpy as np
```

```python
import cv2
import matplotlib.pyplot as plt

# Read and convert images
veg = cv2.imread('veg.png')
veg = cv2.cvtColor(veg, cv2.COLOR_BGR2RGB)
image = cv2.imread('chama.png')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Create subplots
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.imshow(veg)
plt.axis('off')  # Remove axes
plt.title('Veggies')

plt.subplot(1, 2, 2)
plt.imshow(image)
plt.axis('off')  # Remove axes
plt.title('Alex periera')

plt.show()
```

```python
B, G, R = cv2.split(image)

# Create zeros array with the same shape as the original image
zeros = np.zeros_like(B)

# Merge arrays to form images of individual channels
red_image = cv2.merge([zeros, zeros, R])
green_image = cv2.merge([zeros, G, zeros])
blue_image = cv2.merge([B, zeros, zeros])

red_image = cv2.cvtColor(red_image, cv2.COLOR_BGR2RGB)
green_image = cv2.cvtColor(green_image, cv2.COLOR_BGR2RGB)
blue_image = cv2.cvtColor(blue_image, cv2.COLOR_BGR2RGB)

# Plot images as subplots
fig, axs = plt.subplots(1, 3, figsize=(15, 5))
axs[0].imshow(red_image)
axs[0].set_title('Red Channel')
axs[0].axis('off')

axs[1].imshow(green_image)
axs[1].set_title('Green Channel')
axs[1].axis('off')

axs[2].imshow(blue_image)
axs[2].set_title('Blue Channel')
axs[2].axis('off')

plt.show()
```
